```python
import pygrib

try:
    grbs = pygrib.open(era5_path_1989)
    print(f"Number of messages: {grbs.messages}")
    for i, grb in enumerate(grbs):
        if i >= 5:
            break
        print(grb)
    grbs.close()
except Exception as e:
    print("Failed to read GRIB file:", e)
```

```python
import rasterio
from rasterio.crs import CRS

# Open the GRIB file
with rasterio.open(era5_path, mode="r+") as src:
    # Print the current CRS for debugging (optional)
    print("Original CRS:", src.crs)

    # Set the CRS to WGS84 (EPSG:4326) if it's not already set correctly
    if src.crs != CRS.from_epsg(4326):
        src.crs = CRS.from_epsg(4326)
        print("CRS updated to EPSG:4326")

    # Changes are saved automatically when using mode='r+'
```
