```python
import os

model = "ACCESS-CM2"
dir_path = "/global/scratch/projects/co_carleton/carleton_colab/data/gdnat_allmodels_Jul24/aggregated/usa_pop_county/ACCESS-CM2"

for filename in os.listdir(dir_path):
    if filename.startswith("gdnat_usa_agg_") and filename.count("_") == 3 and filename.endswith(".csv"):
        # Expected original format: gdnat_usa_agg_{year}.csv
        parts = filename.split("_")
        # parts example: ['gdnat', 'usa', 'agg', '{year}.csv']
        year_part = parts[-1]  # '{year}.csv'
        new_filename = f"gdnat_usa_agg_{model}_{year_part}"
        src = os.path.join(dir_path, filename)
        dst = os.path.join(dir_path, new_filename)
        print(f"Renaming:\n  {filename} --> {new_filename}")
        os.rename(src, dst)
```
