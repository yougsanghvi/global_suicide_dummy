```python
import os
import pandas as pd
import sys

sys.path.append("/global/home/users/yougsanghvi/global_suicide_dummy/code/y_utils")
import config

def combine_csv_files_by_year(folder, file_format, start_year, end_year):
    """
    Combines CSV files from a given folder using a filename pattern like 'prefix_{year}.csv'
    and saves the combined file as 'prefix_STARTYEAR_ENDYEAR.csv'.

    Parameters:
        folder (str): Path to the folder containing the CSV files.
        file_format (str): Filename pattern with '{year}' placeholder, e.g., 'climate_data_{year}.csv'.
        start_year (int): Starting year (inclusive).
        end_year (int): Ending year (inclusive).
    """
    dfs = []
    for year in range(start_year, end_year + 1):
        filename = file_format.format(year=year)
        filepath = os.path.join(folder, filename)
        if os.path.exists(filepath):
            print(f"Reading {filepath}")
            dfs.append(pd.read_csv(filepath))
        else:
            print(f"Warning: File not found: {filepath}")
    
    if not dfs:
        print("No files were loaded. Exiting.")
        return

    combined_df = pd.concat(dfs, ignore_index=True)

    # Extract prefix before {year} to construct output filename
    prefix = file_format.split("{year}")[0].rstrip('_')
    output_filename = f"{prefix}_{start_year}_{end_year}.csv"
    output_path = os.path.join(folder, output_filename)

    combined_df.to_csv(output_path, index=False)
    print(f"Saved combined file to {output_path}")
```

```python
models = ["MIROC6", "ACCESS-ESM1-5", "CanESM5", "MRI-ESM2-0", "NorESM2-LM", "ACCESS-CM2"]

for model in models:
    folder_path = os.path.join(config.GDNAT_USA_AGG_ALL_MODELS_FP, model)
    file_format = f"gdnat_usa_agg_{model}_{{year}}.csv"
    combine_csv_files_by_year(
        folder=folder_path,
        file_format=file_format,
        start_year=1979,
        end_year=2004
    )
```
